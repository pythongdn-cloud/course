## Zadanie 1. Generator kodów dostępowych

Firma chce wdrożyć pilotażowy system generowania losowych kodów dostępowych dla swoich pracowników. Przygotuj program, który będzie automatycznie generował takie kody.

### Założenia:

- W firmie pracuje 30 osób.
- Każdy pracownik otrzymuje jeden kod dziennie.
- Kody składają się z 6 cyfr.
- Kody nie mogą się powtarzać w całym okresie pilotażowym.
- Program działa przez 30 dni.

### Dodatkowe wymagania:

- Po zakończeniu działania programu wyświetl liczbę wygenerowanych kodów.
- Sprawdź, czy wszystkie wygenerowane kody są unikalne.


## Zadanie 2. Odliczanie do urodzin

Przygotuj program, który sprawdzi, ile dni pozostało do najbliższych urodzin użytkownika.

### Wymagania:

1. Zapoznaj się z modułem `datetime` i zaimportuj go do programu.
2. Pobierz aktualną datę za pomocą `date.today()`.
3. Poproś użytkownika o podanie dnia i miesiąca urodzin.
4. Na podstawie podanych informacji utwórz datę najbliższych urodzin.
5. Oblicz różnicę między aktualną datą a datą urodzin.
6. Wyświetl komunikat informujący, ile dni pozostało do urodzin.

### Dodatkowe wymaganie:

- Jeżeli urodziny użytkownika w bieżącym roku już minęły, program powinien uwzględnić datę z kolejnego roku
