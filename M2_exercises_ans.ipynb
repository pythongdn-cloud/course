**3. Twoja kolej! Użyj pętli while.**

Mamy listę z następującymi odczytami temperatury: 36.6 40.0 39.7 35.1 37.5 36.0

Sprawdź dla każdej temperatury, czy jest wyższa niż 37.0 stopni Celcjusza. Stwórz i wyświetl nową listę, która będzie przechowywać **stan pacjenta**, czyli wartości tekstowe : gorączka, brak gorączki dla każdej z podanych temperatur.

```python
temperatury = [36.6, 40.0, 39.7, 35.1, 37.5, 36.0]
stan_pacjenta = []
licznik = 0
while licznik < len(temperatury): #warunek można zapisać też: licznik <= len(temperatury)-1
  if temperatury[licznik] > 37.0:
    stan_pacjenta.append("gorączka")
  else:
    stan_pacjenta.append("brak gorączki")
  licznik+=1
print(stan_pacjenta)
```

**6. Twoja kolej!**

Oblicz ile razy w słowie rabarbar występuje litera r.

```python
word = "Rabarbar"
count = 0
for znak in word:
  if znak.lower() == "r":
    count+=1
print(f"W słowie {word} są {count} litery \"r\".")

#Pokaż szybszą metodę :)

ile_r = word.lower().count("r")
print("Liczba 'r' znaków to:", ile_r)
```

**7. Twoja kolej! Użyj pętli for.**

```python
zarobki_przed = [9000,10000,9500,10500]
zarobki_po = []

for pensja in zarobki_przed:
  zarobki_po.append(pensja*1.1)
print(zarobki_po)
```

**12. Twoja kolej! Kontrola budżetu działów firmy.**

Pracujesz w dziale finansowym. Firma przydzieliła budżety poszczególnym działom i chcesz sprawdzić, które z nich przekroczyły limit wydatków. Mamy dane 2 słowniki, **budzet** oraz **wydatki**.

Polecenia:
1. Wyświetl każdy dział i jego budżet.
2. Dla każdego działu oblicz różnicę między budżetem a wydatkami.
3. Wyświetl działy, które przekroczyły budżet.
4. Oblicz łączny budżet wszystkich działów.
5. Oblicz łączne wydatki firmy.
6. Wyświetl informację, czy firma zmieściła się w łącznym budżecie. Jeśli nie, to wyświetl informację, że firma się nie zmieściła.

```python
budzet = {
    "IT": 50000,
    "Marketing": 30000,
    "HR": 15000,
    "Logistyka": 40000,
    "Sprzedaz": 25000
}

wydatki = {
    "IT": 47000,
    "Marketing": 33500,
    "HR": 12000,
    "Logistyka": 42500,
    "Sprzedaz": 22000
}
```

```python
#1. Wyświetl każdy dział i jego budżet.

for dzial, budzet_dzialu in budzet.items():
  print(f"Dział: {dzial}, Budżet: {budzet_dzialu:,}".replace(","," "))
  #:, - dodaje przecinek jako separator tysięcy
  #.replace(","," ") - zmienia przecinek na spację
```

```python
#2. Dla każdego działu oblicz różnicę między budżetem a wydatkami.
for dzial in budzet.keys(): #albo for dzial in budzet:
  roznica = budzet[dzial]-wydatki[dzial]
  print(dzial, roznica)
```

```python
#3. Wyświetl działy, które przekroczyły budżet.
for dzial in budzet.keys(): #albo for dzial in budzet:
  roznica = budzet[dzial]-wydatki[dzial]
  if roznica < 0:
    print(dzial)
```

```python
#4. Oblicz łączne wydatki firmy.
łączne_wydatki = sum(wydatki.values())
print(f"{łączne_wydatki:,}".replace(","," "))
```

```python
#5. Oblicz łączny budżet wszystkich działów.
łączny_budżet = sum(budzet.values())
print(f"{łączny_budżet:,}".replace(","," "))
```

```python
#6. Wyświetl informację, czy firma zmieściła się w łącznym budżecie.
if łączne_wydatki <= łączny_budżet:
  print("Firma zmieściła się w budżecie")
else:
  print("Wydatki za duże. Wiadomo co robić! Tnij koszty!")
```
