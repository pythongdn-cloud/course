**3. Twoja kolej! Użyj pętli while.**

Mamy listę z następującymi odczytami temperatury: 36.6 40.0 39.7 35.1 37.5 36.0

Sprawdź dla każdej temperatury, czy jest wyższa niż 37.0 stopni Celcjusza. Stwórz i wyświetl nową listę, która będzie przechowywać **stan pacjenta**, czyli wartości tekstowe : gorączka, brak gorączki dla każdej z podanych temperatur.

```python
temperatury = [36.6, 40.0, 39.7, 35.1, 37.5, 36.0]
stan_pacjenta = []
i = 0 # tu trzeba urzyc zmiennej pomocniczej zeby przejsc po elementach list

while i < len(temperatury): # przechodzimy po elemantach listy
        if temperatury[i] > 37.0: # czyli koknretn temeratura w indeksie krory sprawdzamy
                stan_pacjenta.append("goraczka")
        else:
                stan_pacjenta.append("brak goraczki")

        i += 1 # dodajemy zeby kolejny element listy mogl byc zbadany

print(stan_pacjenta)
print(temperatury)
```

**12. Twoja kolej! Kontrola budżetu działów firmy.**

Pracujesz w dziale finansowym. Firma przydzieliła budżety poszczególnym działom i chcesz sprawdzić, które z nich przekroczyły limit wydatków. Mamy dane 2 słowniki, **budzet** oraz **wydatki**.

Polecenia:
1. Wyświetl każdy dział i jego budżet.
2. Dla każdego działu oblicz różnicę między budżetem a wydatkami.
3. Wyświetl działy, które przekroczyły budżet.
4. Oblicz łączny budżet wszystkich działów.
5. Oblicz łączne wydatki firmy.
6. Wyświetl informację, czy firma zmieściła się w łącznym budżecie. Jeśli nie, to wyświetl informację, że firma się nie zmieściła.

```python
budzet = {
    "IT": 50000,
    "Marketing": 30000,
    "HR": 15000,
    "Logistyka": 40000,
    "Sprzedaz": 25000
}

wydatki = {
    "IT": 47000,
    "Marketing": 33500,
    "HR": 12000,
    "Logistyka": 42500,
    "Sprzedaz": 22000
}

for dzial, pieniadze in budzet.items():
    print(dzial, pieniadze) # wyswietlam dzial i budzet


for dzial, pieniadze in budzet.items():
    roznica = pieniadze - wydatki[dzial]
    print(dzial, roznica)

for dzial, pieniadze in budzet.items():
    if wydatki[dzial] > pieniadze:
        print(dzial) # te dzialy nie zmiesily sie w budzecie 


suma_budzetu = sum(budzet.values()) # suma budzetu 
suma_wydatkow = sum(wydatki.values()) 

if suma_budzetu > suma_wydatkow:
    print('Firma zmiescila sie w budzecie')
else:
    print('Firma nie zmiescila sie w budzecie')
    

```
